````python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import time
import os

WORK_DIR = "/content/drive/MyDrive/심화캡스톤라벨링"  # Drive 마운트 권장
os.makedirs(WORK_DIR, exist_ok=True)

CRAWL_SAVE_PATH = f"{WORK_DIR}/crawled_articles.csv"

df = pd.read_excel(f"{WORK_DIR}/1.xlsx")

url_col = None
for col in ["URL", "링크", "url", "link", "뉴스 식별자"]:
    if col in df.columns:
        url_col = col
        break
print(f"URL 컬럼: {url_col}, 총 기사: {len(df)}건")

def crawl_article(url):
    try:
        headers = {"User-Agent": "Mozilla/5.0"}
        resp = requests.get(url, headers=headers, timeout=10)

        # ✅ 인코딩 강제 지정
        resp.encoding = resp.apparent_encoding  # 먼저 시도
        text_content = resp.text

        # apparent_encoding이 깨지면 utf-8, cp949 순으로 fallback
        if "â€" in text_content or "ì" in text_content:
            for enc in ["utf-8", "cp949", "euc-kr"]:
                try:
                    text_content = resp.content.decode(enc)
                    break
                except:
                    continue

        soup = BeautifulSoup(text_content, "html.parser")
        selectors = [
            "div#articleBodyContents", "div.article_body",
            "div#article-view-content-div", "div#articeBody",
            "div.news_view", "div#newsct_article",
            "article", "div.article-body", "div.story-news",
        ]
        for sel in selectors:
            content = soup.select_one(sel)
            if content and len(content.get_text(strip=True)) > 100:
                return content.get_text(strip=True)
        paragraphs = soup.find_all("p")
        text = " ".join([p.get_text(strip=True) for p in paragraphs])
        return text if len(text) > 100 else None
    except:
        return None

# ✅ 이어하기: 저장된 파일 있으면 불러오기
if os.path.exists(CRAWL_SAVE_PATH):
    df_saved = pd.read_csv(CRAWL_SAVE_PATH)
    df["full_content"] = df_saved["full_content"] if "full_content" in df_saved.columns else None
    print(f"✅ 체크포인트 발견! 기존 {df['full_content'].notna().sum()}건 유지, 이어서 시작")
else:
    df["full_content"] = None
    print("새로 시작")

for i, row in df.iterrows():
    # 이미 크롤링된 행 스킵
    if pd.notna(df.at[i, "full_content"]):
        continue

    url = row[url_col] if url_col else None
    if url and str(url).startswith("http"):
        df.at[i, "full_content"] = crawl_article(str(url))
        time.sleep(0.5)

    if (i + 1) % 10 == 0:
        success = df["full_content"].notna().sum()
        print(f"진행: {i+1}/{len(df)} (성공: {success}건)")

    # 💾 200건마다 저장
    if (i + 1) % 200 == 0:
        df.to_csv(CRAWL_SAVE_PATH, index=False, encoding="utf-8-sig")
        print(f"  💾 체크포인트 저장! ({i+1}건)")

# 최종 저장
df.to_csv(CRAWL_SAVE_PATH, index=False, encoding="utf-8-sig")
print(f"\n✅ 크롤링 완료 및 저장! 성공: {df['full_content'].notna().sum()}건")
````

````python
# ============================================================
# 📌 셀 5: LLM 보조 라벨링 함수 정의
# ============================================================

import json
import time
import pandas as pd


LABELED_SAVE_PATH = f"{WORK_DIR}/labeled_sentences.csv"

# ✅ 이어하기
if os.path.exists(LABELED_SAVE_PATH):
    df_sentences = pd.read_csv(LABELED_SAVE_PATH)
    print(f"✅ 체크포인트 발견! {len(df_sentences)}건 불러옴")
else:
    if "df_sentences" not in dir():
        df_sentences = pd.read_csv(SAMPLED_SAVE_PATH)
    print("새로 시작")


def label_sentence(sentence, task="emotion"):
    """
    GPT-4o-mini로 문장 하나를 라벨링합니다.

    task 종류:
    - emotion: 감정 편향 (0=neutral, 1=biased_negative, 2=biased_positive)
    - fact_opinion: 사실/의견 (0=fact, 1=opinion)
    - sensational: 선정성 (0=normal, 1=sensational)
    """

    if task == "emotion":
        prompt = f"""다음 뉴스 문장의 감정 편향을 분류해주세요.

문장: "{sentence}"

분류 기준:
- 0 (neutral): 감정적 표현 없이 사실만 전달. 형용사/부사를 빼도 의미가 같은 문장.
- 1 (biased_negative): 부정적 감정 표현 포함. "무능한", "뻔뻔한", "파탄" 등.
- 2 (biased_positive): 긍정적 감정 표현 포함. "탁월한", "훌륭한", "눈부신" 등.

반드시 아래 JSON 형식으로만 답해주세요. 다른 텍스트 없이 JSON만:
{{"label": 0, "confidence": 0.9, "reason": "판단 근거 한 줄"}}"""

    elif task == "fact_opinion":
        prompt = f"""다음 뉴스 문장이 사실 서술인지 의견 서술인지 분류해주세요.

문장: "{sentence}"

분류 기준:
- 0 (fact): 객관적으로 검증 가능한 사실. 수치, 날짜, 공식 발표 등.
- 1 (opinion): 주관적 판단, 평가, 주장. "~해야 한다", "~라고 볼 수 있다" 등.

반드시 아래 JSON 형식으로만 답해주세요. 다른 텍스트 없이 JSON만:
{{"label": 0, "confidence": 0.9, "reason": "판단 근거 한 줄"}}"""

    elif task == "sensational":
        prompt = f"""다음 뉴스 문장이 선정적인지 판별해주세요.

문장: "{sentence}"

분류 기준:
- 0 (normal): 일반적인 뉴스 표현. 차분하고 객관적.
- 1 (sensational): 자극적, 과장, 감정 호소. "충격", "경악", "폭탄", "분노 폭발" 등.

반드시 아래 JSON 형식으로만 답해주세요. 다른 텍스트 없이 JSON만:
{{"label": 0, "confidence": 0.9, "reason": "판단 근거 한 줄"}}"""

    try:
        response = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.1,
            max_tokens=150,
        )
        content = response.choices[0].message.content.strip()

        # JSON 파싱 (가끔 ```json ... ``` 형태로 오는 경우 처리)
        content = content.replace("```json", "").replace("```", "").strip()
        result = json.loads(content)

        return result.get("label", -1), result.get("confidence", 0), result.get("reason", "")

    except json.JSONDecodeError:
        return -1, 0, f"JSON 파싱 실패: {content[:50]}"
    except Exception as e:
        return -1, 0, f"API 에러: {str(e)[:50]}"


# 테스트
test_label, test_conf, test_reason = label_sentence(
    "정부의 무능한 경제 정책이 서민 경제를 파탄으로 몰아가고 있다.", "emotion"
)
print(f"테스트 결과: label={test_label}, confidence={test_conf}, reason={test_reason}")


# ============================================================
# 📌 셀 6: 전체 라벨링 실행 (3개 태스크)
# ============================================================
# 3,000건 기준 약 10~15분 소요, 비용 약 $0.05
# ============================================================

# ✅ 이어하기: 저장된 파일 있으면 불러오기
if os.path.exists(LABELED_SAVE_PATH):
    df_sentences = pd.read_csv(LABELED_SAVE_PATH)
    print(f"✅ 체크포인트 발견! {len(df_sentences)}건 불러옴")
else:
    if "df_sentences" not in dir():
        df_sentences = pd.read_csv(SAMPLED_SAVE_PATH)
    print("새로 시작")

tasks = ["emotion", "fact_opinion", "sensational"]
task_names = {
    "emotion":      "감정 편향",
    "fact_opinion": "사실/의견",
    "sensational":  "선정성",
}

for task in tasks:
    label_col = f"{task}_label"

    # ✅ 이미 완료된 task 스킵
    if label_col in df_sentences.columns:
        done = pd.to_numeric(df_sentences[label_col], errors="coerce").notna().sum()
        if done == len(df_sentences):
            print(f"⏭️ {task_names[task]} 이미 완료 ({done}건), 스킵")
            continue

    print(f"\n{'='*60}")
    print(f"🏷️ {task_names[task]} 라벨링 시작 ({len(df_sentences)}건)")
    print(f"{'='*60}")

    # 컬럼 없으면 초기화
    for col in [f"{task}_label", f"{task}_confidence", f"{task}_reason"]:
        if col not in df_sentences.columns:
            df_sentences[col] = None

    for i, row in df_sentences.iterrows():
        # ✅ 이미 라벨링된 행 스킵
        if pd.notna(df_sentences.at[i, label_col]):
            continue

        label, conf, reason = label_sentence(row["sentence"], task)
        df_sentences.at[i, f"{task}_label"]      = label
        df_sentences.at[i, f"{task}_confidence"] = conf
        df_sentences.at[i, f"{task}_reason"]     = reason

        time.sleep(0.05)

        if (i + 1) % 100 == 0:
            done = pd.to_numeric(df_sentences[label_col], errors="coerce").notna().sum()
            print(f"  진행: {done}/{len(df_sentences)} ({done/len(df_sentences)*100:.0f}%)")

        # 💾 500건마다 중간 저장
        if (i + 1) % 500 == 0:
            df_sentences.to_csv(LABELED_SAVE_PATH, index=False, encoding="utf-8-sig")
            print(f"  💾 체크포인트 저장!")

    # human_label 컬럼 추가
    if f"{task}_human_label" not in df_sentences.columns:
        df_sentences[f"{task}_human_label"] = ""
        df_sentences[f"{task}_verified"]    = False

    # task 완료 후 저장
    df_sentences.to_csv(LABELED_SAVE_PATH, index=False, encoding="utf-8-sig")

    valid = pd.to_numeric(df_sentences[label_col], errors="coerce").dropna()
    print(f"\n  완료: {len(valid)}건 성공, {len(df_sentences)-len(valid)}건 실패")
    print(f"  라벨 분포: {valid.astype(int).value_counts().sort_index().to_dict()}")
    print(f"  💾 저장 완료!")

print(f"\n🎉 전체 라벨링 완료!")
````

````python
# ============================================================
# 📌 셀 7: 결과 저장 (엑셀 + CSV)
# ============================================================

# 엑셀 저장 (팀원 검수용 — 엑셀이 편함)
excel_path = f"{WORK_DIR}/labeled_data_for_review.xlsx"
df_sentences.to_excel(excel_path, index=False)
print(f"📁 엑셀 저장: {excel_path}")

# CSV 저장 (학습 코드용)
csv_path = f"{WORK_DIR}/labeled_data.csv"
df_sentences.to_csv(csv_path, index=False, encoding="utf-8-sig")
print(f"📁 CSV 저장: {csv_path}")

# 태스크별 개별 CSV도 저장 (학습 시 바로 사용)
for task in tasks:
    task_df = df_sentences[["sentence", f"{task}_label", f"{task}_confidence", f"{task}_reason"]].copy()
    task_df = task_df.rename(columns={f"{task}_label": "label", f"{task}_confidence": "confidence", f"{task}_reason": "reason"})
    task_df = task_df[task_df["label"] >= 0]  # 실패한 건 제외

    task_path = f"{WORK_DIR}/labeled_{task}.csv"
    task_df.to_csv(task_path, index=False, encoding="utf-8-sig")
    print(f"📁 {task_names[task]} CSV: {task_path} ({len(task_df)}건)")

print(f"\n✅ 모든 파일 저장 완료!")
print(f"📂 저장 경로: {WORK_DIR}")
````

````python
# ============================================================
# 📌 셀 9: 검수 가이드 출력
# ============================================================

print("""
╔══════════════════════════════════════════════════════════════╗
║                    📋 팀원 검수 가이드                        ║
╠══════════════════════════════════════════════════════════════╣
║                                                              ║
║  1. Drive에서 labeled_data_for_review.xlsx 다운로드           ║
║                                                              ║
║  2. 각 태스크별로 검수:                                       ║
║     - emotion_label 확인 → emotion_human_label에 값 입력      ║
║     - fact_opinion_label 확인 → fact_opinion_human_label 입력 ║
║     - sensational_label 확인 → sensational_human_label 입력   ║
║                                                              ║
║  3. LLM 라벨에 동의하면 같은 값 입력                          ║
║     동의하지 않으면 올바른 값 입력                             ║
║                                                              ║
║  4. 검수 완료한 행의 verified 컬럼을 TRUE로 변경              ║
║                                                              ║
║  5. 완료된 엑셀을 Drive에 다시 업로드                         ║
║                                                              ║
╠══════════════════════════════════════════════════════════════╣
║  감정 편향: 0=중립, 1=부정편향, 2=긍정편향                    ║
║  사실/의견: 0=사실, 1=의견                                    ║
║  선정성:   0=보통, 1=선정적                                   ║
╠══════════════════════════════════════════════════════════════╣
║                                                              ║
║  💡 팁: LLM이 이미 라벨+근거를 제시해주니까                   ║
║     "동의/비동의"만 판단하면 됩니다.                          ║
║     1인당 750건 기준 약 1시간 소요.                           ║
║                                                              ║
╚══════════════════════════════════════════════════════════════╝
""")

# 팀원별 분배
total = len(df_sentences)
team_size = 4
per_person = total // team_size

print(f"팀원 분배 ({total}건 / {team_size}명 = 1인당 {per_person}건):")
print(f"  재원:   1번 ~ {per_person}번")
print(f"  팀원A:  {per_person+1}번 ~ {per_person*2}번")
print(f"  팀원B:  {per_person*2+1}번 ~ {per_person*3}번")
print(f"  팀원C:  {per_person*3+1}번 ~ {total}번")
````

````python
# ============================================================
# 📌 셀 10: 검수 완료 후 — 최종 학습 데이터 생성
# ============================================================
# 팀원들이 검수 완료한 엑셀을 Drive에 올린 후 실행하세요
# ============================================================

REVIEWED_FILE = f"{WORK_DIR}/labeled_data_for_review.xlsx"  # 검수 완료 파일

if os.path.exists(REVIEWED_FILE):
    df_reviewed = pd.read_excel(REVIEWED_FILE)

    for task in ["emotion", "fact_opinion", "sensational"]:
        human_col = f"{task}_human_label"
        llm_col = f"{task}_label"
        verified_col = f"{task}_verified"

        # 검수 완료된 것만
        verified = df_reviewed[df_reviewed[verified_col] == True]

        if len(verified) == 0:
            print(f"⚠️ {task}: 검수 완료된 데이터 없음")
            continue

        # 최종 라벨: human_label이 있으면 human, 없으면 llm
        final_labels = []
        for _, row in verified.iterrows():
            if pd.notna(row[human_col]) and row[human_col] != "":
                final_labels.append(int(row[human_col]))
            else:
                final_labels.append(int(row[llm_col]))

        # LLM-사람 일치율
        agreement_count = sum(
            1 for _, row in verified.iterrows()
            if pd.notna(row[human_col]) and row[human_col] != "" and int(row[human_col]) == int(row[llm_col])
        )
        has_human = sum(1 for _, row in verified.iterrows() if pd.notna(row[human_col]) and row[human_col] != "")

        if has_human > 0:
            agreement = agreement_count / has_human * 100
            print(f"\n{task}: LLM-사람 일치율 = {agreement:.1f}% ({agreement_count}/{has_human})")

        # 최종 학습 데이터 저장
        final_df = pd.DataFrame({
            "text": verified["text"].values,
            "label": final_labels,
        })
        final_df = final_df[final_df["label"] >= 0]

        final_path = f"{WORK_DIR}/final_{task}.csv"
        final_df.to_csv(final_path, index=False, encoding="utf-8-sig")
        print(f"📁 최종 {task} 데이터: {final_path} ({len(final_df)}건)")
        print(f"   라벨 분포: {final_df['label'].value_counts().sort_index().to_dict()}")

    print(f"\n✅ 최종 학습 데이터 생성 완료!")
    print(f"이 CSV 파일들로 klue/roberta-base 파인튜닝을 진행하면 됩니다.")

else:
    print(f"⚠️ 검수 완료 파일이 없습니다: {REVIEWED_FILE}")
    print("팀원 검수 완료 후 다시 실행해주세요.")
````
